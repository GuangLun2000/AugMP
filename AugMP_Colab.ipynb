## Step 5: Downstream explanation generation (optional)

Task 2: the **SeqCLS head** classifies each probe, then the **CausalLM** (backbone transferred from the same checkpoint) generates an explanation. Label space: `0=World, 1=Sports, 2=Business, 3=Sci/Tech`. Output format: `Category: ...\nReason: ...`. Supported backbones: **Qwen2 / Qwen2.5** and **Pythia / GPT-NeoX** (see `decoder_adapters.py`).


```python
if globals().get('RUN_MAIN_EXPERIMENT', True):
    from pathlib import Path
    import json
    import subprocess
    import sys

    import torch

    RESULTS = Path("results")
    EXPERIMENT_NAME = "vgae_augmp"
    CKPT_SUBDIR = "global_checkpoint"
    summary = RESULTS / f"{EXPERIMENT_NAME}_results.json"
    if summary.exists():
        with open(summary, encoding="utf-8") as f:
            data = json.load(f)
            cfg = data.get("config", {})
            EXPERIMENT_NAME = cfg.get("experiment_name", EXPERIMENT_NAME)
            CKPT_SUBDIR = cfg.get("global_checkpoint_subdir", CKPT_SUBDIR)

    ckpt_dir = RESULTS / CKPT_SUBDIR
    pt = ckpt_dir / "global_model.pt"
    meta_path = ckpt_dir / "checkpoint_metadata.json"


    def print_all_probe_pairs_from_jsonl(jsonl_path: Path) -> None:
        """Print every probe: SeqCLS classification, dataset label, and CausalLM explanation."""
        if not jsonl_path.is_file():
            print(f"\n(no JSONL to dump: {jsonl_path} not found)\n")
            return
        with open(jsonl_path, encoding="utf-8") as f:
            lines = [ln for ln in f if ln.strip()]
        print("\n" + "=" * 60)
        print(f"ALL probe results — {len(lines)} rows from {jsonl_path}")
        print("=" * 60)
        for idx, line in enumerate(lines, start=1):
            row = json.loads(line)
            pid = row.get("probe_id", idx)
            news = row.get("news_text", "")
            print(f"\n{'#' * 12} Probe id={pid} ({idx}/{len(lines)}) {'#' * 12}\n")
            print("[INPUT] news_text:\n", news, sep="")
            if "seq_cls_category_id" in row:
                print("\n[SeqCLS] category_id / category:\n",
                      row.get("seq_cls_category_id"), " / ", row.get("seq_cls_category"), sep="")
            if "dataset_label_id" in row or "dataset_category" in row:
                print("\n[DATASET] category_id / category:\n",
                      row.get("dataset_label_id"), " / ", row.get("dataset_category"), sep="")
            print("\n[OUTPUT] completion:\n", row.get("completion", ""), sep="")
            if "reason_raw" in row:
                print("\n[DEBUG] reason_raw:\n", row.get("reason_raw"), sep="")
            print("\n[JSON row pretty]:\n", json.dumps(row, ensure_ascii=False, indent=2), sep="")
            print()
        print("=" * 60)


    print("=" * 60)
    print("Global checkpoint (Fed → downstream)")
    print("=" * 60)
    if pt.exists():
        print(f"✅ Found: {pt}")
        if meta_path.exists():
            with open(meta_path, encoding="utf-8") as f:
                meta = json.load(f)
            print(json.dumps(meta, indent=2))
    else:
        print(f"⚠️ No checkpoint at {pt}")
        print("   Set save_global_checkpoint=True in main.py or COLAB_CONFIG_OVERRIDES, then re-run Step 3.")

    out_jsonl = RESULTS / "colab_downstream_gen.jsonl"
    probes_path = Path("data/AG News Datasets/ag_news_business_30.json")

    if not pt.exists():
        pass
    elif not probes_path.exists():
        print(f"⚠️ Probes missing: {probes_path}")
    else:
        model_name = ""
        if meta_path.exists():
            with open(meta_path, encoding="utf-8") as f:
                model_name = json.load(f).get("model_name", "")
        low = model_name.lower()
        supported = ("pythia" in low or "gpt-neox" in low or "qwen2" in low)
        if not supported:
            print(
                f"⚠️ model_name={model_name!r} — no decoder adapter for this checkpoint.\n"
                "   Use EleutherAI/pythia-160m or Qwen/Qwen2.5-* (see decoder_adapters.py)."
            )
        else:
            device_arg = "cuda" if torch.cuda.is_available() else "cpu"
            cmd = [
                sys.executable,
                "run_downstream_generation.py",
                "--checkpoint", str(ckpt_dir),
                "--probes", str(probes_path),
                "--output", str(out_jsonl),
                "--stable",
                "--device", device_arg,
            ]
            print("Running:", " ".join(cmd))
            proc = subprocess.run(cmd, capture_output=True, text=True)
            print(proc.stdout)
            if proc.stderr:
                print(proc.stderr)
            if proc.returncode != 0:
                print(f"Exit code: {proc.returncode}")

    print_all_probe_pairs_from_jsonl(out_jsonl)
else:
    print('Skipped (RUN_MAIN_EXPERIMENT = False)')
```

## Step 6: Observability experiment (attackers observe only *k* benign updates)

Runs live in `observe/` and write `results/obs_*_results.json`; figures and the CSV summary go to `results/observe/`.
All training/attack hyper-parameters are pinned inside `observe/run_observability_sweep.py` (DistilBERT, AG News, 10 agents = 7 benign + 3 attackers, Dirichlet 0.3, LoRA r=8); only the observation settings change between runs.
The first cell runs the sweep (each entry is one GPU run; finished runs are skipped, so re-running the cell after a disconnect resumes). The second cell embeds the figures and the summary table **inside this notebook's outputs**, so they stay visible after the runtime is released, and offers a zip download of the figures, CSV and raw JSON results.

```python
# Observability sweep. Each entry is one GPU run in a fresh subprocess; runs whose results already exist are
# skipped (--skip-existing), so re-running this cell after a disconnect simply resumes.
import subprocess, sys

ROUNDS_ANCHOR = 50   # full-observation anchors (benign, ALIE, AugMP k=7); paper standard
ROUNDS_SWEEP = 30    # limited-observation runs; compared with the anchors at round 30

STAGE_1 = [  # core: anchors + headline point + main line (largest-data peers + global-broadcast anchor)
    f"--attack none ALIE AugMP --k 7 --rounds {ROUNDS_ANCHOR}",
    f"--k 4 2 --rounds {ROUNDS_SWEEP}",
    f"--k 2 --mode largest --anchor --rounds {ROUNDS_SWEEP}",
]
STAGE_2 = [  # optional: more levels, a second observed subset, ALIE under the same limitation, other strategies
    f"--k 3 --rounds {ROUNDS_SWEEP}",
    f"--k 2 --obs-seed 1 --rounds {ROUNDS_SWEEP}",
    f"--attack ALIE --k 2 --rounds {ROUNDS_SWEEP}",
    f"--k 2 --mode largest --rounds {ROUNDS_SWEEP}",
    f"--k 2 --anchor --rounds {ROUNDS_SWEEP}",
    f"--k 2 --mode adaptive --criterion deviant --anchor --rounds {ROUNDS_SWEEP}",
    f"--k 0 --rounds {ROUNDS_SWEEP}",   # Secure Aggregation endpoint: no benign updates observed, only the global broadcast
    # frozen-bound ceiling: first `python observe/plot_observability.py --suggest-bounds --no-plots`, then
    # f"--k 2 --rounds {ROUNDS_SWEEP} --dist-bound <dT> --sim-bound-up <s> --suffix fixedbounds",
]
RUN_STAGES = [STAGE_1]   # append STAGE_2 to run it as well

if globals().get('RUN_OBSERVE_EXPERIMENT', True):
    # Optional: keep results on Drive (uncomment before the first run)
    # from google.colab import drive; drive.mount('/content/drive')
    # subprocess.run("mkdir -p /content/drive/MyDrive/augmp_obs_results && (test -e results || ln -s /content/drive/MyDrive/augmp_obs_results results)", shell=True, check=True)
    for stage in RUN_STAGES:
        for flags in stage:
            cmd = f"{sys.executable} observe/run_observability_sweep.py {flags} --skip-existing"
            print("\n>>>", cmd, flush=True)
            rc = subprocess.run(cmd, shell=True).returncode
            if rc != 0:
                print(f"!!! run failed (exit {rc}); see results/observe/logs/", flush=True)
    print("\nObservability sweep finished.")
else:
    print("Skipped (RUN_OBSERVE_EXPERIMENT = False)")
```
